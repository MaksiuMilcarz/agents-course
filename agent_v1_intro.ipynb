```python
"""
Hugging Face smolagents + oMLX Starter
======================================
Connects a local smolagents runner to an oMLX OpenAI-compatible endpoint serving:
Qwen3.5-9B-Claude-4.6-HighIQ-INSTRUCT-HERETIC-UNCENSORED-mlx-4Bit
"""

import math
from smolagents import CodeAgent, DuckDuckGoSearchTool, OpenAIServerModel, tool, load_tool

# 1. Server & Model Configuration
# oMLX exposes standard OpenAI-compatible completions/chat endpoints.
MODEL_ID = "Qwen3.5-9B-Claude-4.6-HighIQ-INSTRUCT-HERETIC-UNCENSORED-mlx-4Bit"
OMLX_BASE_URL = "http://localhost:8000/v1"  # Default oMLX API prefix
OMLX_API_KEY = ""

model = OpenAIServerModel(
    model_id=MODEL_ID,
    api_base=OMLX_BASE_URL,
    api_key=OMLX_API_KEY,  # oMLX runs locally without auth checks
    temperature=0.2,       # Lower temperature improves code/tool-call determinism
)

image_generation_tool = load_tool("agents-course/text-to-image", trust_remote_code=True)
```

```python
# 2. Custom Tool Definition
# smolagents uses typed function signatures and docstrings to inform the agent's planner.
@tool
def calculate_compound_growth(principal: float, annual_rate: float, years: int) -> str:
    """Calculates future value using continuous compound growth: A = P * e^(r*t).

    Args:
        principal: Starting financial amount or initial base value.
        annual_rate: Annual growth rate as a decimal (e.g. 0.07 for 7%).
        years: Total integer years elapsed.

    Returns:
        str: Summary string containing the calculated final balance.
    """
    total = principal * math.exp(annual_rate * years)
    return f"Calculated future value after {years} years: {total:,.2f}"
```

```python
# 3. Agent Initialization
# CodeAgent generates and executes sandboxed Python code snippets to accomplish tasks.
agent = CodeAgent(
    tools=[DuckDuckGoSearchTool(), calculate_compound_growth, image_generation_tool],
    model=model,
    add_base_tools=True,   # Includes core utility tools like web inspection and time
    max_steps=7,           # Upper bound to guard against infinite plan iterations
    verbosity_level=1,     # Prints thought traces and executed code blocks to stdout
)
```

```python
task = (
    "Calculate the 10-year yield of a 5,000 CHF investment growing at an estimated "
    "6.5% annual rate, then provide a brief one-sentence recap."
)

print(f"Executing task with local oMLX model: {MODEL_ID}\n")
response = agent.run(task)
print("\n--- Final Agent Answer ---")
print(response)
```
